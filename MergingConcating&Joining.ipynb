```python
import numpy as np
import pandas as pd
```

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105],
    "name": ["Rahul", "Amit", "Priya", "Neha", "Rohit"],
    "city": ["Bhopal", "Delhi", "Mumbai", "Bhopal", "Indore"]
})

marks = pd.DataFrame({
    "student_id": [101, 102, 103, 106, 107],
    "subject": ["Python", "Python", "Python", "Python", "Python"],
    "marks": [85, 92, 78, 88, 75]
})
```

```python
students
```

```python
marks
```

```python
# Merging two tables 
```

```python
pd.merge(students,marks, on="student_id",how="inner")
```

```python
pd.merge(students,marks, on="student_id",how="left")
```

```python
pd.merge(students,marks, on="student_id",how="right")
```

```python
pd.merge(students,marks, on="student_id",how="outer")
```

```python
# Concatination of two tables
```

```python
pd.concat([students,marks])
```

```python
pd.concat([students,marks], ignore_index=True)
```

```python
pd.concat([students,marks], axis=1) # concating horizontally
```
