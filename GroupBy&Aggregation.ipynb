```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
```

```python
df=pd.DataFrame(data)
df
```

```python
#Group by category and calcute the sum of the sales
cat=df.groupby('Category')['Sales'].sum()
cat
```

```python
#group by store and calcute the sum of the sales
cat=df.groupby('Store')['Sales'].sum()
cat
```

```python
#average Sales for each Category.
cat=df.groupby('Category')['Sales'].mean()
cat
```

```python
#maximum Sales for each Store.
cat=df.groupby('Store')['Sales'].max()
cat
```

```python
#Total Quantity for each Category and Store.
cat=df.groupby(['Category','Store'])['Quantity'].sum()
cat
```
